```python
combined = pd.concat(
    [
        physical,
        chemical.drop(columns=["_id", "SlNo", "Station", "Agency",
                               "State LGD Code", "State",
                               "District LGD Code", "District",
                               "Tehsil", "Block", "Village",
                               "River", "Basin", "Tributary",
                               "Subtributary", "SubSubtributary",
                               "Local River", "Latitude", "Longitude",
                               "Data Acquisition Time"]),
        biological.drop(columns=["SlNo", "Station", "Agency",
                                 "State LGD Code", "State",
                                 "District LGD Code", "District",
                                 "Tehsil", "Block", "Village",
                                 "River", "Basin", "Tributary",
                                 "Subtributary", "SubSubtributary",
                                 "Local River", "Latitude", "Longitude",
                                 "Data Acquisition Time"])
    ],
    axis=1
)

print(combined.shape)
```

```python
combined["Fecal Coliform (MPN/100mL)"] = pd.to_numeric(
    combined["Fecal Coliform (MPN/100mL)"],
    errors="coerce"
)

print(combined["Fecal Coliform (MPN/100mL)"].dtype)
```

```python
print(combined["Fecal Coliform (MPN/100mL)"].notna().sum())
```

```python
model_data = combined[
    combined["Fecal Coliform (MPN/100mL)"].notna()
].copy()

print(model_data.shape)
```

```python
model_data[
    [
        "Potential of Hydrogen (pH)",
        "Dissolved oxygen (mg/L)",
        "Biochemical Oxygen Demand (mg/L)"
    ]
].isna().sum()
```

```python
core_features = [
    "Potential of Hydrogen (pH)",
    "Dissolved oxygen (mg/L)",
    "Biochemical Oxygen Demand (mg/L)"
]

core_complete = model_data[core_features].notna().all(axis=1)

print("Rows with all core features:", core_complete.sum())
```

```python
baseline_data = model_data.loc[
    core_complete,
    core_features + ["Fecal Coliform (MPN/100mL)"]
].copy()

print(baseline_data.shape)
```

```python
baseline_data["Potential of Hydrogen (pH)"] = pd.to_numeric(
    baseline_data["Potential of Hydrogen (pH)"],
    errors="coerce"
)

baseline_data["Dissolved oxygen (mg/L)"] = pd.to_numeric(
    baseline_data["Dissolved oxygen (mg/L)"],
    errors="coerce"
)

print(baseline_data.dtypes)
```

```python
print(baseline_data.isna().sum())
```

```python
baseline_data = baseline_data.dropna().copy()

print(baseline_data.shape)
```

```python
baseline_data.corr()
```

```python
baseline_data["log_fecal_coliform"] = np.log1p(
    baseline_data["Fecal Coliform (MPN/100mL)"]
)

print(
    baseline_data[
        [
            "Potential of Hydrogen (pH)",
            "Dissolved oxygen (mg/L)",
            "Biochemical Oxygen Demand (mg/L)",
            "log_fecal_coliform"
        ]
    ].corr()["log_fecal_coliform"]
)
```

```python
import matplotlib.pyplot as plt

plt.scatter(
    baseline_data["Biochemical Oxygen Demand (mg/L)"],
    baseline_data["log_fecal_coliform"],
    alpha=0.4
)

plt.xlabel("BOD (mg/L)")
plt.ylabel("Log Fecal Coliform")
plt.title("BOD vs Fecal Coliform")
plt.show()
```

```python
plt.scatter(
    baseline_data["Potential of Hydrogen (pH)"],
    baseline_data["log_fecal_coliform"],
    alpha=0.4
)

plt.xlabel("pH")
plt.ylabel("Log Fecal Coliform")
plt.title("pH vs Fecal Coliform")
plt.show()
```

```python
plt.scatter(
    baseline_data["Dissolved oxygen (mg/L)"],
    baseline_data["log_fecal_coliform"],
    alpha=0.4
)

plt.xlabel("Dissolved Oxygen (mg/L)")
plt.ylabel("Log Fecal Coliform")
plt.title("Dissolved Oxygen vs Fecal Coliform")
plt.show()
```

```python
X = baseline_data[
    [
        "Potential of Hydrogen (pH)",
        "Dissolved oxygen (mg/L)",
        "Biochemical Oxygen Demand (mg/L)"
    ]
]

y = baseline_data["log_fecal_coliform"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```
